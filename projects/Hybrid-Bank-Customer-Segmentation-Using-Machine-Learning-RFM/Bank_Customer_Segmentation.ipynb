# Hybrid Bank Customer Segmentation

Customer-level behavioral segmentation with KMeans, DBSCAN, PCA, and RFM. This notebook is intentionally portable and contains no committed dataset or saved outputs.

## Data and reproducibility

Set `BANK_TRANSACTIONS_PATH` to the CSV location before running. The notebook also checks `bank_transactions.csv` and `data/bank_transactions.csv` under the current working directory. The dataset is not present in this repository, so the load cell raises a clear error until a file is supplied.

This is unsupervised analysis: there is no target and therefore no supervised train/test split. Silhouette values are in-sample descriptive diagnostics. For a production predictive use case, fit preprocessing on an earlier time window and evaluate on a later holdout.

```python
import os
from pathlib import Path

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.cluster import DBSCAN, KMeans
from sklearn.decomposition import PCA
from sklearn.impute import SimpleImputer
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler

RANDOM_STATE = 42
rng = np.random.default_rng(RANDOM_STATE)
np.random.seed(RANDOM_STATE)
sns.set_theme(style="whitegrid")

configured_path = os.getenv("BANK_TRANSACTIONS_PATH")
candidates = []
if configured_path:
    candidates.append(Path(configured_path).expanduser())
candidates.extend([Path.cwd() / "bank_transactions.csv", Path.cwd() / "data" / "bank_transactions.csv"])
DATA_PATH = next((candidate for candidate in candidates if candidate.is_file()), None)
```

```python
if DATA_PATH is None:
    checked = "\n".join(f"  - {candidate}" for candidate in candidates)
    raise FileNotFoundError(
        "Bank transaction data was not found. Set BANK_TRANSACTIONS_PATH to an existing CSV "
        "or place bank_transactions.csv in the repository/data directory. Checked:\n" + checked
    )

df = pd.read_csv(DATA_PATH)
required_columns = {
    "CustomerID",
    "TransactionID",
    "TransactionAmount (INR)",
    "CustAccountBalance",
    "CustomerDOB",
    "TransactionDate",
}
if df.columns.duplicated().any():
    duplicated_columns = df.columns[df.columns.duplicated()].tolist()
    raise ValueError(f"Dataset contains duplicate column names: {duplicated_columns}")
missing_columns = sorted(required_columns.difference(df.columns))
if missing_columns:
    raise ValueError(f"Dataset is missing required columns: {missing_columns}")

print(f"Loaded {len(df):,} rows from {DATA_PATH}")
print("Columns:", list(df.columns))
```

## Data quality and cleaning

Key identifiers, transaction amount, and transaction date are required for aggregation. Rows missing those fields are excluded, and the exclusion count is reported. Optional demographic/time columns are used only when present. No synthetic rows or metric results are created.

```python
raw_rows = len(df)
df = df.drop_duplicates().copy()
exact_duplicates_removed = raw_rows - len(df)

for column in ["TransactionAmount (INR)", "CustAccountBalance"]:
    df[column] = pd.to_numeric(df[column], errors="coerce")
    df.loc[~np.isfinite(df[column]), column] = np.nan

for column in ["CustomerID", "TransactionID"]:
    df[column] = df[column].astype("string").str.strip()

df["CustomerDOB"] = pd.to_datetime(df["CustomerDOB"], dayfirst=True, errors="coerce")
df["TransactionDate"] = pd.to_datetime(df["TransactionDate"], dayfirst=True, errors="coerce")
key_columns = ["CustomerID", "TransactionID", "TransactionAmount (INR)", "TransactionDate"]
blank_identifier = df[["CustomerID", "TransactionID"]].eq("").fillna(False).any(axis=1)
missing_required = df[key_columns].isna().any(axis=1)
missing_key_rows = int((missing_required | blank_identifier).sum())
df = df.loc[~blank_identifier].dropna(subset=key_columns).copy()
if df.empty:
    raise ValueError("No valid transaction rows remain after required-field validation.")

reference_date = df["TransactionDate"].max()
df["Age"] = (reference_date - df["CustomerDOB"]).dt.days / 365.25
df.loc[~df["Age"].between(18, 100), "Age"] = np.nan

if "CustGender" in df:
    df["CustGender"] = df["CustGender"].astype("string").str.strip().replace({"M": "Male", "F": "Female"})
    df.loc[~df["CustGender"].isin(["Male", "Female"]), "CustGender"] = pd.NA
if "CustLocation" in df:
    df["CustLocation"] = df["CustLocation"].astype("string").str.strip().str.upper()
if "TransactionTime" in df:
    time_number = pd.to_numeric(df["TransactionTime"].astype("string").str.zfill(6), errors="coerce")
    df["Hour"] = (time_number // 10000).where(time_number.between(0, 235959))

print(f"Exact duplicate rows removed: {exact_duplicates_removed:,}")
print(f"Rows missing required fields removed: {missing_key_rows:,}")
print(f"Usable rows: {len(df):,}; customers: {df['CustomerID'].nunique():,}")
```

## Customer-level feature engineering

All clustering features are built from transaction history up to the observed reference date. Identifiers are trimmed and transaction frequency counts distinct TransactionID values. Demographics are aggregated within customer, preventing a many-to-one merge from duplicating customers.

```python
def first_non_null(values):
    values = values.dropna()
    return values.iloc[0] if not values.empty else pd.NA


def zero_balance_ratio(values):
    values = values.dropna()
    return float(values.eq(0).mean()) if not values.empty else np.nan


aggregations = {
    "TotalTransactions": ("TransactionID", "nunique"),
    "TotalAmount": ("TransactionAmount (INR)", "sum"),
    "AvgTransactionAmount": ("TransactionAmount (INR)", "mean"),
    "MaxTransactionAmount": ("TransactionAmount (INR)", "max"),
    "AvgAccountBalance": ("CustAccountBalance", "mean"),
    "ZeroBalanceRatio": ("CustAccountBalance", zero_balance_ratio),
    "LastTransactionDate": ("TransactionDate", "max"),
    "FirstTransactionDate": ("TransactionDate", "min"),
    "Age": ("Age", "median"),
}
if "CustGender" in df:
    aggregations["CustGender"] = ("CustGender", first_non_null)

customer_df = df.groupby("CustomerID", as_index=False).agg(**aggregations)
customer_df["Recency"] = (reference_date - customer_df["LastTransactionDate"]).dt.days
customer_df["CustomerLifetimeDays"] = (
    customer_df["LastTransactionDate"] - customer_df["FirstTransactionDate"]
).dt.days.clip(lower=1)
customer_df["Age"] = customer_df["Age"].where(customer_df["Age"].between(18, 100))
customer_df.head()
```

```python
FEATURE_COLUMNS = [
    "TotalTransactions",
    "TotalAmount",
    "AvgTransactionAmount",
    "AvgAccountBalance",
    "ZeroBalanceRatio",
    "Recency",
    "CustomerLifetimeDays",
    "Age",
]
features = customer_df[FEATURE_COLUMNS].replace([np.inf, -np.inf], np.nan).copy()

# Signed log keeps the transformation defined even if refunds or corrections are negative.
for column in ["TotalAmount", "AvgTransactionAmount", "AvgAccountBalance"]:
    features[column] = np.sign(features[column]) * np.log1p(features[column].abs())

imputer = SimpleImputer(strategy="median")
features_imputed = imputer.fit_transform(features)
scaler = StandardScaler()
scaled_features = scaler.fit_transform(features_imputed)

# This is an unsupervised descriptive fit on the full observed population, not a claim of holdout performance.
pca = PCA(n_components=0.95, svd_solver="full")
pca_features = pca.fit_transform(scaled_features)
print(f"Original features: {scaled_features.shape[1]}; PCA components: {pca_features.shape[1]}")
```

## KMeans and DBSCAN

The random seed is fixed. KMeans is fitted on all customer rows. DBSCAN is sampled for runtime control with a seeded generator; only sampled rows receive DBSCAN labels, and unsampled rows are not treated as noise. Silhouette is reported only when its label cardinality is mathematically valid.

```python
if len(pca_features) < 4:
    raise ValueError("At least four customers are required for the configured KMeans model.")


def guarded_silhouette(features, labels, name):
    labels = np.asarray(labels)
    unique_labels = np.unique(labels)
    if not 1 < unique_labels.size < len(labels):
        print(f"{name} silhouette is undefined: need 2 to n-1 distinct labels; got {unique_labels.size} labels for {len(labels)} rows.")
        return np.nan
    return silhouette_score(features, labels)

k_values = range(2, min(10, len(pca_features) - 1) + 1)
inertia = []
for k in k_values:
    model = KMeans(n_clusters=k, random_state=RANDOM_STATE, n_init=10)
    inertia.append(model.fit(pca_features).inertia_)

plt.figure(figsize=(8, 5))
plt.plot(list(k_values), inertia, marker="o")
plt.xlabel("Number of clusters")
plt.ylabel("Inertia")
plt.title("KMeans elbow diagnostic")
plt.show()
```

```python
N_CLUSTERS = 4
kmeans = KMeans(n_clusters=N_CLUSTERS, random_state=RANDOM_STATE, n_init=10)
customer_df["KMeans_Cluster"] = kmeans.fit_predict(pca_features)
kmeans_labels = customer_df["KMeans_Cluster"].to_numpy()
kmeans_silhouette = guarded_silhouette(pca_features, kmeans_labels, "KMeans")
if np.isfinite(kmeans_silhouette):
    print(f"KMeans silhouette (in-sample): {kmeans_silhouette:.4f}")
```

```python
sample_size = min(5_000, len(pca_features))
sample_indices = rng.choice(len(pca_features), size=sample_size, replace=False)
sample_features = pca_features[sample_indices]
dbscan = DBSCAN(eps=0.5, min_samples=20)
dbscan_labels = dbscan.fit_predict(sample_features)
non_noise = dbscan_labels != -1
sample_non_noise_features = sample_features[non_noise]
sample_non_noise_labels = dbscan_labels[non_noise]
dbscan_silhouette = guarded_silhouette(
    sample_non_noise_features, sample_non_noise_labels, "DBSCAN sample"
)
if np.isfinite(dbscan_silhouette):
    print(f"DBSCAN sample: {sample_size:,} rows; non-noise silhouette: {dbscan_silhouette:.4f}")
else:
    print(f"DBSCAN sample: {sample_size:,} rows; silhouette is undefined for the sampled non-noise rows.")
```

## RFM segmentation

RFM scores are calculated from customer-level recency, frequency, and monetary values. Ranking before `qcut` makes ties deterministic and avoids duplicate-bin errors.

```python
rfm = customer_df[["CustomerID", "Recency", "TotalTransactions", "TotalAmount"]].rename(
    columns={"Recency": "R", "TotalTransactions": "F", "TotalAmount": "M"}
).copy()

def quartile_score(series, reverse=False):
    rank = series.rank(method="first")
    quantiles = min(4, len(series))
    labels = list(range(1, quantiles + 1))
    if reverse:
        labels.reverse()
    return pd.qcut(rank, q=quantiles, labels=labels).astype(int)

rfm["R_Score"] = quartile_score(rfm["R"], reverse=True)
rfm["F_Score"] = quartile_score(rfm["F"])
rfm["M_Score"] = quartile_score(rfm["M"])
rfm["RFM_Segment"] = rfm[["R_Score", "F_Score", "M_Score"]].astype(str).agg("".join, axis=1)

def rfm_segment_label(row):
    if row["R_Score"] >= 3 and row["F_Score"] >= 3 and row["M_Score"] >= 3:
        return "VIP"
    if row["R_Score"] >= 3 and row["F_Score"] >= 2:
        return "Loyal"
    if row["R_Score"] <= 2 and row["F_Score"] >= 2:
        return "At Risk"
    if row["R_Score"] <= 2 and row["F_Score"] <= 2:
        return "Lost"
    return "Regular"

rfm["RFM_Segment_Label"] = rfm.apply(rfm_segment_label, axis=1)
rfm.head()
```

## Interpretation boundaries

Cluster labels are arbitrary identifiers, not business meanings. Interpret segments only after inspecting `cluster_profile` and the underlying feature distributions. The notebook does not assert that a particular cluster is VIP, dormant, or risky without data-backed profiling. Metrics shown after execution are diagnostics, not externally validated results.
